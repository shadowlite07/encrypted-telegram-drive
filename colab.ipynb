````python
%%writefile /content/telestore-colab.js
const { createDrive, loadConfig } = require('encrypted-telegram-drive');
const fs = require('node:fs');
const path = require('node:path');

async function main() {
  const configPath = process.env.CONFIG_PATH || '/content/config.json';
  const config = loadConfig(configPath);
  const drive = createDrive(config, process.env.DRIVE_KEY, { quiet: true });

  const health = await drive.doctor();
  console.log(JSON.stringify(health, null, 2));

  const files = drive.files.list({ all: true });
  console.log('Stored files:', files.length);

  drive.close();
}

main().catch((error) => {
  console.error(error);
  process.exit(1);
});
````
